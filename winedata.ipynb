```python
# loc helps to find item by focusing labels.
wine_reviews.loc[0,['country']]
```

```python
wine_reviews.loc[1:5,['country','designation','winery']]
```

```python
# If iloc uses slicing like 1:5 it includes 1 to 4 excludes 5.
# if loc uses slicing like 1:5 it includes 1 to 5 all.
```
