```python
newdf = pd.DataFrame({
    "name":['Alfred','Batman','Alfred'],
    "toy":[np.nan,np.nan,np.nan],
    "born":[pd.NaT,pd.Timestamp("1940-04-25"),pd.NaT]
})
```

```python
newdf
```

```python
newdf.dropna()
```

```python
newdf.dropna(how='all') #doesnot drop until all records are na
```

```python
newdf.dropna(how='all',axis=1)
```

```python
newdf.drop_duplicates()
```

```python
newdf.drop_duplicates(subset=['name'])
```

```python
newdf.drop_duplicates(subset=['name'],keep='last')
```

```python
newdf.drop_duplicates(subset=['name'],keep=False)
```
