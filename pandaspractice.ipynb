## Merging Dataframe

### We can merge dataframes based on common column by using "on" argument.

```python

data1 = {'key': ['K0', 'K1', 'K2', 'K3'],
         'Name':['Jai', 'Princi', 'Gaurav', 'Anuj'],
        'Age':[27, 24, 22, 32],}

data2 = {'key': ['K0', 'K1', 'K2', 'K3'],
         'Address':['Nagpur', 'Kanpur', 'Allahabad', 'Kannuaj'],
        'Qualification':['Btech', 'B.A', 'Bcom', 'B.hons']}

df = pd.DataFrame(data1)

df1 = pd.DataFrame(data2)


print(df, "\n\n", df1)
```

```python
mer_result=pd.merge(df,df1,on='key') #It merges on the basis of key column
mer_result
```

```python
import pandas as pd

data1 = {'key': ['K0', 'K1', 'K2', 'K3'],
         'key1': ['K0', 'K1', 'K0', 'K1'],
         'Name':['Jai', 'Princi', 'Gaurav', 'Anuj'],
        'Age':[27, 24, 22, 32],}

data2 = {'key': ['K0', 'K1', 'K2', 'K3'],
         'key1': ['K0', 'K0', 'K0', 'K0'],
         'Address':['Nagpur', 'Kanpur', 'Allahabad', 'Kannuaj'],
        'Qualification':['Btech', 'B.A', 'Bcom', 'B.hons']}

df = pd.DataFrame(data1)

df1 = pd.DataFrame(data2)


print(df, "\n\n", df1)
```

```python
res4=pd.merge(df,df1,on=['key','key1']) #It merges on the basis of multiple column and gives only the data whose keys match
res4
```

```python
# We can join using how method, left outer,right outer,full outer,inner join
```

```python
res5 = pd.merge(df,df1,how='left',on=['key','key1'])
res5
```

### Data Reshaping functions (pivot table() and melt())

#### PIVOT TABLE

```python
df4 = pd.DataFrame({
    'Name': ['Ram', 'Ram', 'Hari', 'Hari'],
    'Subject': ['Math', 'Science', 'Math', 'Science'],
    'Marks': [90, 85, 80, 75]
})

print(df4)
```

```python
table = df4.pivot_table(
    index='Name',
    columns='Subject',
    values='Marks'
    
)
```

```python
table
```

```python
df5 = pd.DataFrame({
    'Name': ['Ram', 'Ram'],
    'Subject': ['Math', 'Math'],
    'Marks': [90, 85,]
})

print(df5)
```

```python
# It returns the aggregate function by default (mean)
# Returned average marks of Ram in Maths
table1=df5.pivot_table(
    index='Name',
    columns='Subject',
    values='Marks'
)
table1
```

```python
# To change aggregate function we have to explicitly change it 
table1=df5.pivot_table(
    index='Name',
    columns='Subject',
    values='Marks',
    aggfunc='sum'
)
table1
```

#### MELT()

```python
df6 = pd.DataFrame({
    'Name': ['Ram', 'Hari'],
    'Math': [90, 80],
    'Science': [85, 75]
})

pd.melt(df6,id_vars='Name')
```
